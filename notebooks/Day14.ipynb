```python
import os
from google import genai
```

```python
api_key = os.environ.get("GEMINI_API_KEY")
```

```python
client = genai.Client(api_key=api_key)
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="Hello Gemini! Give me a quick one-sentence greeting.",
)

print(response.text)
```

```python
import json

sample_reviews = [
    "Battery life is amazing, easily lasts 2 days of heavy use. But the screen scratches way too easily.",
    "Audio quality on these headphones is crisp, though they feel a bit tight on larger heads after an hour.",
    "Super lightweight and comfortable. Noise cancellation works great on flights, but the mic picks up too much wind.",
    "Stopped charging after 3 weeks. Customer support was slow to respond. Very disappointed."
]

def summarize_product_reviews(product_name: str, reviews: list[str]) -> str:
    reviews_text = "\n".join([f"- {r}" for r in reviews])
    
    prompt = f"""
    You are an expert e-commerce analytics assistant. 
    Analyze the following customer reviews for '{product_name}' and provide a concise summary broken down into:
    1. Key Pros
    2. Key Cons
    3. Overall Verdict / Theme
    
    Reviews:
    {reviews_text}
    """
    
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt,
    )
    return response.text

summary_output = summarize_product_reviews("Wireless Audio Pods", sample_reviews)
print("=== Review Summarization ===")
print(summary_output)
```

```python

product_policy_context = """
[Product Metadata]
- Name: Wireless Audio Pods
- Price: $129.99
- Battery Life: Up to 30 hours total with charging case (6 hours continuous playback).
- Water Resistance: IPX4 splash-proof.

[Return & Support Policy]
- Return Window: 30 days from the date of purchase with original receipt.
- Warranty: 1-year limited hardware warranty covering manufacturing defects.
- Support Hours: Monday-Friday, 9 AM - 5 PM EST.
"""

chat = client.chats.create(
    model="gemini-3.5-flash-lite",
    config=genai.types.GenerateContentConfig(
        system_instruction=(
            "You are a helpful and professional customer support chatbot for an e-commerce brand. "
            "You must answer user questions strictly based on the provided product metadata and support policies. "
            "If the answer cannot be found in the text, politely tell the user you don't know and invite them to contact support during working hours."
            f"\n\nContext:\n{product_policy_context}"
        ),
        temperature=0.2,
    )
)
```

```python
response1 = chat.send_message("What is your return policy and warranty length?")
print("User: What is your return policy and warranty length?")
print(f"Chatbot:\n{response1.text}\n")
```

```python
response2 = chat.send_message("How long does the battery last on a single charge?")
print("User: How long does the battery last on a single charge?")
print(f"Chatbot:\n{response2.text}\n")
```
