```python
import pandas as pd
from sklearn.model_selection import train_test_split
import joblib
```

```python
MODEL_PATH="../models/churn_pipeline.joblib"
URL="../data/Telco-Customer-Churn.csv"
```

```python
def predict(data: pd.DataFrame) -> dict:
    """
    Thin ML Service wrapper to load the persisted pipeline and output churn predictions.
    """
    pipeline = joblib.load(MODEL_PATH)
    
    # Generate binary prediction & churn probabilities
    predictions = pipeline.predict(data)
    probabilities = pipeline.predict_proba(data)[:, 1]

    return {
        'predictions': predictions.tolist(),
        'churn_probabilities': probabilities.tolist()
    }
```

```python
df=pd.read_csv(URL)
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
    
    
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].astype(str).str.strip(), errors='coerce')
    
    
X = df.drop(columns=['customerID', 'Churn'])
y = df['Churn']
Xtest,Xtrain,ytrain,ytest =train_test_split(X,y,test_size=0.20)
res=predict(X[:1])
```
